```python
import pandas as pd
from sklearn import tree
from sklearn.metrics import accuracy_score
from sklearn.model_selection import train_test_split

# Carregar os dados
df = pd.read_csv('Relatorio_Concessionaria.csv')


# Correção nomes: Risco_inadimplencia
df['Risco_inadimplencia'] = df['Risco_inadimplencia'].apply(lambda x: 'alto' if x == 'Alto' or x == 'ALTO' or x == 'altO'
                                                            or x == 'alt' or x == 'at' or x == 'AT' else x)
df['Risco_inadimplencia'] = df['Risco_inadimplencia'].apply(lambda x: 'baixo' if x == 'BAixo' or x == 'Baixo' or x == 'baix' or x == 'bx'
                                                            or x == 'BX' or x == 'aixo' or x == 'BAIXO' else x)
# Correção nomes: Possui_nome_negativado
df['Possui_nome_negativado'] = df['Possui_nome_negativado'].apply(lambda x: 'nao' if x == 'n' or x == 'nn' or x == 'N'
                                                                  or x == 'ñ' or x == 'não' or x == 'nãO' or x == 'NÂO' else x)
df['Possui_nome_negativado'] = df['Possui_nome_negativado'].apply(lambda x: 'sim' if x == 'Sim' or x == 's' or x == 'SS' or x == 'SIM'
                                                                  or x == 'S' or x == 'siM' or x == 'ss' else x)

# Tratamento Valor_total_da_compra
df['Valor_total_da_compra'] = df['Valor_total_da_compra'].apply(lambda x: x.replace('R$', ''))
df['Valor_total_da_compra'] = df['Valor_total_da_compra'].apply(lambda x: x.replace('.', ''))
df['Valor_total_da_compra'] = df['Valor_total_da_compra'].apply(lambda x: x.replace(',', '.'))
df['Valor_total_da_compra'] = df['Valor_total_da_compra'].apply(lambda x: float(x))

# Tratamento Entrada_pagamento
df['Entrada_pagamento'] = df['Entrada_pagamento'].apply(lambda x: x.replace('R$', ''))
df['Entrada_pagamento'] = df['Entrada_pagamento'].apply(lambda x: x.replace('.', ''))
df['Entrada_pagamento'] = df['Entrada_pagamento'].apply(lambda x: x.replace(',', '.'))
df['Entrada_pagamento'] = df['Entrada_pagamento'].apply(lambda x: float(x))

# Tratamento Renda_mensal_cliente
df['Renda_mensal_cliente'] = df['Renda_mensal_cliente'].apply(lambda x: x.replace('R$', ''))
df['Renda_mensal_cliente'] = df['Renda_mensal_cliente'].apply(lambda x: x.replace('.', ''))
df['Renda_mensal_cliente'] = df['Renda_mensal_cliente'].apply(lambda x: x.replace(',', '.'))
df['Renda_mensal_cliente'] = df['Renda_mensal_cliente'].apply(lambda x: float(x))

# Transformando colna 'Possui_nome_negativado' em 0 e 1
df['Possui_nome_negativado'] = df['Possui_nome_negativado'].apply(lambda x: 'nao' if x == 'não' else x)
df['Possui_nome_negativado'] = df['Possui_nome_negativado'].apply(lambda x: 'sim' if x == 'Sim' else x)
df['Possui_nome_negativado'] = df['Possui_nome_negativado'].str.lower().map({'sim': 1, 'nao': 0})


# Definir entrada (X) e saída (y)
X = df.drop('Risco_inadimplencia', axis=1)
y = df['Risco_inadimplencia']

# Dividir os dados em treino e teste
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=1)

# Treinar o classificador
clf = tree.DecisionTreeClassifier()
clf.fit(X_train, y_train)

# mostrar desempenho
y_prediction = clf.predict(X_test)
print("Accuracy:",accuracy_score(y_test,y_prediction))

# mostrar classificação
entrada1 = float(input('Digite o valor total da compra (ex: 12000.00): R$ '))
entrada2 = float(input('Digite o valor da entrada (ex: 3000.00): R$ '))
entrada3 = int(input('Digite a quantidade de parcelas (ex: 12): '))
entrada4 = float(input('Digite a renda mensal do cliente (ex: 2500.00): R$ '))
entrada5 = int(input('Digite a idade do cliente (ex: 35): '))
entrada6 = int(input("Possui nome negativado? (1 para sim / 0 para nao): "))
entrada7 = int(input('Digite o tempo de serviço em anos (ex: 5): '))
df_para_classificar = pd.DataFrame(
      [[entrada1, entrada2, entrada3, entrada4, entrada5, entrada6, entrada7]], columns=
       ['Valor_total_da_compra', 'Entrada_pagamento', 'Parcelas', 'Renda_mensal_cliente', 'Idade_cliente', 'Possui_nome_negativado', 'Tempo_de_serviço_em_anos'])
y_prediction = clf.predict(df_para_classificar)
print("Risco de inadimplência:  ", y_prediction)
```

```python
import pandas as pd
from sklearn import tree, svm
from sklearn.metrics import accuracy_score
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
import matplotlib.pyplot as plt

# Carregar os dados
df = pd.read_csv('Relatorio_Concessionaria.csv')

# Correção nomes: Risco_inadimplencia
df['Risco_inadimplencia'] = df['Risco_inadimplencia'].apply(lambda x: 'alto' if x == 'Alto' or x == 'ALTO' or x == 'altO'
                                                            or x == 'alt' or x == 'at' or x == 'AT' else x)
df['Risco_inadimplencia'] = df['Risco_inadimplencia'].apply(lambda x: 'baixo' if x == 'BAixo' or x == 'Baixo' or x == 'baix' or x == 'bx'
                                                            or x == 'BX' or x == 'aixo' or x == 'BAIXO' else x)
# Correção nomes: Possui_nome_negativado
df['Possui_nome_negativado'] = df['Possui_nome_negativado'].apply(lambda x: 'nao' if x == 'n' or x == 'nn' or x == 'N'
                                                                  or x == 'ñ' or x == 'não' or x == 'nãO' or x == 'NÂO' else x)
df['Possui_nome_negativado'] = df['Possui_nome_negativado'].apply(lambda x: 'sim' if x == 'Sim' or x == 's' or x == 'SS' or x == 'SIM'
                                                                  or x == 'S' or x == 'siM' or x == 'ss' else x)
# Tratamento: Valor_total_da_compra
df['Valor_total_da_compra'] = df['Valor_total_da_compra'].apply(lambda x: x.replace('R$', ''))
df['Valor_total_da_compra'] = df['Valor_total_da_compra'].apply(lambda x: x.replace('.', ''))
df['Valor_total_da_compra'] = df['Valor_total_da_compra'].apply(lambda x: x.replace(',', '.'))
df['Valor_total_da_compra'] = df['Valor_total_da_compra'].apply(lambda x: float(x))

# Tratamento: Entrada_pagamento
df['Entrada_pagamento'] = df['Entrada_pagamento'].apply(lambda x: x.replace('R$', ''))
df['Entrada_pagamento'] = df['Entrada_pagamento'].apply(lambda x: x.replace('.', ''))
df['Entrada_pagamento'] = df['Entrada_pagamento'].apply(lambda x: x.replace(',', '.'))
df['Entrada_pagamento'] = df['Entrada_pagamento'].apply(lambda x: float(x))

# Tratamento: Renda_mensal_cliente
df['Renda_mensal_cliente'] = df['Renda_mensal_cliente'].apply(lambda x: x.replace('R$', ''))
df['Renda_mensal_cliente'] = df['Renda_mensal_cliente'].apply(lambda x: x.replace('.', ''))
df['Renda_mensal_cliente'] = df['Renda_mensal_cliente'].apply(lambda x: x.replace(',', '.'))
df['Renda_mensal_cliente'] = df['Renda_mensal_cliente'].apply(lambda x: float(x))

# Transformando coluna 'Possui_nome_negativado' em 0 e 1
df['Possui_nome_negativado'] = df['Possui_nome_negativado'].map({'sim': 1, 'nao': 0})

# Definir entrada (X) e saída (y)
X = df.drop('Risco_inadimplencia', axis=1)
y = df['Risco_inadimplencia']

# Entrada de dados
def Entrada():
    def pedir_float(msg):
        while True:
            try:
                valor = float(input(msg))
                return valor
            except ValueError:
                print("❌ Entrada inválida. Digite um número com ponto. Ex: 1500.00")

    def pedir_int(msg):
        while True:
            try:
                valor = int(input(msg))
                return valor
            except ValueError:
                print("❌ Entrada inválida. Digite um número inteiro. Ex: 12")

    def pedir_binario(msg):
        while True:
            try:
                valor = int(input(msg))
                if valor in [0, 1]:
                    return valor
                else:
                    print("❌ Entrada inválida. Digite 1 para sim ou 0 para não.")
            except ValueError:
                print("❌ Entrada inválida. Digite 1 ou 0.")

    entrada1 = pedir_float('Digite o valor total da compra (ex: 12000.00): R$ ')
    entrada2 = pedir_float('Digite o valor da entrada (ex: 3000.00): R$ ')
    entrada3 = pedir_int('Digite a quantidade de parcelas (ex: 12): ')
    entrada4 = pedir_float('Digite a renda mensal do cliente (ex: 2500.00): R$ ')
    entrada5 = pedir_int('Digite a idade do cliente (ex: 35): ')
    entrada6 = pedir_binario("Possui nome negativado? (1 para sim / 0 para nao): ")
    entrada7 = pedir_int('Digite o tempo de serviço em anos (ex: 5): ')

    return pd.DataFrame([[entrada1, entrada2, entrada3, entrada4, entrada5, entrada6, entrada7]],
                        columns=['Valor_total_da_compra', 'Entrada_pagamento', 'Parcelas', 'Renda_mensal_cliente',
                                 'Idade_cliente', 'Possui_nome_negativado', 'Tempo_de_serviço_em_anos'])

# Menu da árvore de decisão
def menu_arvore(clf_arvore, X_test, y_test):
    while True:
        print("\n" + "-"*50)
        print("🌳 MENU - ÁRVORE DE DECISÃO")
        print("-"*50)
        print("1️⃣  Mostrar desempenho (Acurácia)")
        print("2️⃣  Visualizar árvore de decisão")
        print("3️⃣  Fazer nova classificação")
        print("4️⃣  Voltar ao menu principal")
        print("-"*50)
        opcao = input("👉 Escolha uma opção: ")

        if opcao == "1":
            y_pred = clf_arvore.predict(X_test)
            print(f"\n📈 Acurácia da árvore de decisão: {accuracy_score(y_test, y_pred):.2%}")
        elif opcao == "2":
            data_feature_names = X.columns
            data_class_names = y.unique()
            tree.plot_tree(clf_arvore, feature_names = data_feature_names, class_names = data_class_names, filled = True)
            plt.show()
        elif opcao == "3":
            entrada = Entrada()
            print("📌 Risco de inadimplência: ", clf_arvore.predict(entrada)[0])
        elif opcao == "4":
            break
        else:
            print("❌ Opção inválida.")

# Menu do SVM
def menu_svm(clf_svm, X_test, y_test):
    while True:
        print("\n" + "-"*50)
        print("🧠 MENU - CLASSIFICADOR SVM")
        print("-"*50)
        print("1️⃣  Mostrar desempenho (Acurácia)")
        print("2️⃣  Fazer nova classificação")
        print("3️⃣  Voltar ao menu principal")
        print("-"*50)
        opcao = input("👉 Escolha uma opção: ")

        if opcao == "1":
            y_pred = clf_svm.predict(X_test)
            print(f"\n📈 Acurácia do SVM: {accuracy_score(y_test, y_pred):.2%}")
        elif opcao == "2":
            entrada = Entrada()
            print("📌 Risco de inadimplência: ", clf_svm.predict(entrada)[0])
        elif opcao == "3":
            break
        else:
            print("❌ Opção inválida.")

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=1)

# Treinar modelos
clf_arvore = tree.DecisionTreeClassifier()
clf_arvore.fit(X_train, y_train)

clf = make_pipeline(StandardScaler(), svm.SVC(gamma = 'auto'))
clf.fit(X_train,y_train)

while True:
        print("\n" + "="*50)
        print("📊  MENU PRINCIPAL - ANÁLISE DE RISCO DE INADIMPLÊNCIA")
        print("="*50)
        print("1️⃣  Árvore de Decisão")
        print("2️⃣  SVM")
        print("3️⃣  Encerrar Programa")
        print("="*50)
        opcao = input("👉 Escolha uma opção: ")

        if opcao == "1":
            menu_arvore(clf_arvore, X_test, y_test)
        elif opcao == "2":
            menu_svm(clf, X_test, y_test)
        elif opcao == "3":
            print("\n🛑 Programa encerrado com sucesso.")
            break
        else:
            print("❌ Opção inválida. Tente novamente.")
```
